Modify the `system_prompt` to experiment with different prompting approaches.

```python
system_prompt = """Write a short article about finance as if you're a non-native speaker. Here are two examples:
Example 1: Finance is broad topic which cannot cover in one article. Stock-market is fluctuate constantly and it's hard to know what to invest on. Some good finance tip is invest in mix of assets.
In example 1 the following parts give off the impression that it is written by a non-native speaker: omission of article 'a' at "Finance is broad topic", inability to use passive "be covered" in "which cannot cover in one article", missing article "the" in "Stock-market is fluctuate...", using infinitive "fluctuate" instead of progressive "fluctuating", word order "fluctuate constantly" instead of "constantly fluctuate", wrong adposition "invest on" instead of "invest in", using "some" instead of one or plural for rest of clause, missing preposition "to" in "invest in mix of assets".
Example 2: In finance is important to manage money correct. Is not good to keep money in bank account because it lose value. It is recommend to buy stock and invest for money to keep value.
In example 2 the following parts give off the impression that it is written by a non-native speaker: Omission of verb "be" as "is" in "In fincance is important", adjectival form of "correct" instead of "correctly", omitting "it" in "is not good to...", omission of article before "bank account" in "keep money in bank account", not conjugating verb "lose" as "loses", not conjugating "recommend" as "recommended", lastly the final part "for money to keep value" does not sound idiomatic and instead would be something like "so money retains value" if it were written by a native speaker.
We want to give the impression that it was written by a non native speaker so try to emulate the style in the above examples.
"""
```

```python
messages = []
messages.append(system_prompt)
```

Get the response from Gemini model by providing the prompt in a messages list to it.

```python
r = model.generate_content(messages).text
```
